## Part 1: SCAD class with real data
The <a href="https://www.kaggle.com/datasets/crawford/80-cereals">cereals</a> dataset, which has nutritional information on various cereal brands, and the SCAD model is used to predict the amount of sugar in different cereals.

```python
# Loading the data
data = pd.read_csv('../data/cereal.csv')
x_df = data.drop(columns=['name', 'mfr', 'type', 'sugars'])
x = x_df.values
y = data['sugars'].values

scaler = MinMaxScaler()
xscaled = scaler.fit_transform(x)

# Once the data has been scaled, we can convert it to a tensor for use with the SCAD model
x_tensor = torch.tensor(xscaled, dtype= torch.float64)
y_tensor = torch.tensor(y, dtype= torch.float64)
```

```python
# Now we can fit the model on the data. Then, we can make predictions on the data and see if the prediction is accurate.
model = scad_linear.SCADLinear(input_size= x_tensor.shape[1], lambda_val= 1, a_val= 6)
# We use tensor.unsqueeze(1) to make the dimensions of x and y equal. Otherwise, there is a broadcasting error which lowers model performance.
model.fit(x= x_tensor, y= y_tensor.unsqueeze(1), num_epochs= 2000, learning_rate= 0.29)
prediction = model.predict(x_tensor)
```

```python
# Comparison with the actual y values shows a fairly low MSE.
mse(y_tensor, prediction)
```

```python
# Getting the coefficients the model is using. It appears that the variable 'vitamins' has a weight of -0.9192, making it not as impactful.
model.get_coefficients()
```

## Part 2: Strong Correlation Structure
This section generates 200 samples of strongly correlated (rho=0.9) data and compares the results using SCAD, ElasticNet, and SqrtLasso.

```python
import numpy as np
from scipy.linalg import toeplitz
import models_for_comparison
import matplotlib.pyplot as plt
```

```python
# Code to make correlated data
def make_correlated(num_samples, p, rho):
    vcor = []
    for i in range(p):
        vcor.append(rho**i)
    r = toeplitz(vcor)
    mu = np.repeat(0,p)
    x = np.random.multivariate_normal(mu, r, size=num_samples)
    return x
```

```python
rho = 0.9 # strongness of the data
p = 7 # number of features in the data
n = 200 # 200 sets of correlated data
x = make_correlated(n, p, rho)
beta =np.array([-1,2,3,0,1,1])
beta = beta.reshape(-1,1)
betastar = np.concatenate([beta,np.repeat(0,p-len(beta)).reshape(-1,1)],axis=0) # fully defined correlation structure
y = x@betastar + 1.5*np.random.normal(size=(n,1))

x_tensor = torch.tensor(x, device=device)
y_tensor = torch.tensor(y, device= device)
```

```python
scad = scad_linear.SCADLinear(x_tensor.shape[1], 0.5, 3) # create each model
elastic = models_for_comparison.ElasticNet(input_size=x_tensor.shape[1], alpha=0.1, l1_ratio=0.6)
sqrt = models_for_comparison.SqrtLasso(input_size= x_tensor.shape[1], alpha=0.1)
```

```python
scad.fit(x_tensor, y_tensor, num_epochs=1000, learning_rate=0.01) # fit each model using identical learning rates for consistency
elastic.fit(x_tensor, y_tensor, num_epochs=1000, learning_rate=0.01)
sqrt.fit(x_tensor, y_tensor, num_epochs=1000, learning_rate= 0.01)
```

```python
# Make a list of all the coefficients that each model has predicted
coefficients = {'Original coefficients': betastar.flatten(),
                'SCADLinear': scad.get_coefficients().detach().numpy().flatten(),
                'SqrtLasso': sqrt.get_coefficients().detach().numpy().flatten(),
                'ElasticNet': elastic.get_coefficients().detach().numpy().flatten()}
coef = pd.DataFrame(coefficients)
pd.set_option('display.float_format', '{:,.3f}'.format)

# Iterate through the scores of each model, including what model they were from
for model in [scad, elastic, sqrt]:
    score = mse(y_tensor.detach().numpy(), model.predict(x_tensor).detach().numpy())
    print(f'model: {model}\nscore: {score}')
```

SCAD has the lowest score on a dataset of 200 data points. SCAD also has most closely predicted the actual correlation structure for most of the variables.

## Part 3: Concrete Dataset
In this section, SCAD predicts the strength of concrete using the concrete dataset, and transforming the input data to represent quadratic terms. Optimal hyperparameters were determined and tested with 10-fold cross-validation.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import KFold
import math
from sklearn.model_selection import GridSearchCV
import itertools
```

```python
data = pd.read_csv('../data/concrete.csv')

x = data.drop(columns='strength').values
y = data['strength'].values

# Transform the data to include the quadratic terms
polynomial = PolynomialFeatures(degree=2, include_bias=False)
x_polynomial = polynomial.fit_transform(x)
scaler = MinMaxScaler()
xscaled = scaler.fit_transform(x_polynomial)

x_tensor = torch.tensor(xscaled, dtype= dtype)
y_tensor = torch.tensor(y, dtype= dtype)
```

```python
lambda_possible_vals = [0.001, 0.01, 0.1]
a_possible_vals = [2.0, 3.0, 4.0]
```

```python
param_grid = list(itertools.product(lambda_possible_vals, a_possible_vals))
scores = []

# Iterating through a list of parameters for a and lambda
for lambda_val, a_val in param_grid:
    model = scad_linear.SCADLinear(input_size = x_tensor.shape[1], lambda_val = lambda_val, a_val = a_val)
    model.fit(x_tensor, y_tensor.unsqueeze(1), num_epochs = 1000, learning_rate = 0.1)
    scores.append(mse(y_tensor, model.predict(x_tensor)))
```

```python
scores # fairly similar scores for each hyperparameter, possibly a result of the learning rate
```

```python
model = scad_linear.SCADLinear(input_size=x_tensor.shape[1], lambda_val= 0.01, a_val= 3) # choosing values for lambda and a that seem to be good
```

```python
model.fit(x_tensor, y_tensor.unsqueeze(1), num_epochs = 1000, learning_rate = 0.1)
```

```python
model.get_coefficients()
```

```python
scores = []
threshold = 0.01 # close to 0 threshold when counting number of coefficients
coefficients_counted = []

kf = KFold(n_splits = 10)
for train_index, test_index in kf.split(x):
    x_train, x_test = x_polynomial[train_index], x_polynomial[test_index]
    y_train, y_test = y[train_index], y[test_index]

    x_train_scale = scaler.fit_transform(x_train)
    x_test_scale = scaler.fit_transform(x_test)

    model.fit(torch.tensor(x_train_scale), torch.tensor(y_train).unsqueeze(1), num_epochs = 1000, learning_rate = 0.2)
    y_predicted = model.predict(torch.tensor(x_test_scale))
    print(mse(y_test, y_predicted.detach().numpy()))

    coef_np = model.get_coefficients().detach().numpy()
    coefficients_counted.append(np.sum(np.abs(coef_np) > threshold))
    
    scores.append(mse(y_test, y_predicted.detach().numpy()))
```

```python
print(f'Cross-validated scores: {scores}')
print(f'Best number of coefficients: {coefficients_counted}')
```

### Figures for part 3

```python
fig, ax = plt.subplots()
ax.scatter(coefficients_counted, scores)
plt.xlabel('Best number of coefficients')
plt.ylabel('Score')
plt.xlim([40, 45])
```
